```python
z=np.random.rand(len(df)) >0.8
train = x[z]
Test = x[~z]
```

```python
plt.scatter(train.ENGINESIZE, train.CO2EMISSIONS, color="blue")
plt.xlabel("ENGINESIZE")
plt.ylabel("EMISSION")
plt.show()
```

```python
from sklearn import linear_model
d= linear_model.LinearRegression()
train_x = np.asanyarray(train[['ENGINESIZE']])
train_y = np.asanyarray(train[['CO2EMISSIONS']])
d.fit(train_x,train_y)

print('coeffication:' , d.coef_)
print('intercept:', d.intercept_)
```

```python
plt.scatter(train.ENGINESIZE, train.CO2EMISSIONS, color="blue")
plt.plot(train_x, d.coef_[0][0]*train_x + d.intercept_[0], '-r')
plt.xlabel("ENGINESIZE")
plt.ylabel("CO2EMISSION")
```

```python

test_x = np.asanyarray(df[['ENGINESIZE']])
test_y = np.asanyarray(df[['CO2EMISSIONS']])
test_y_ = d.predict(test_x)

print("Mean absolute error: %.2f" % np.mean(np.absolute(test_y_ - test_y)))
print("Residual sum of squares (MSE): %.2f" % np.mean((test_y_ - test_y) ** 2))
print("R2-score: %.2f" % r2_score(test_y , test_y_) )
```
