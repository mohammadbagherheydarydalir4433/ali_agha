```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df=pd.read_csv("classification_moons.csv")
df.head()
```

```python
X=df[['x1' , 'x2']].values
X[0:5]
```

```python
y=df['label'].values
y[0:5]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size=0.2 , random_state=4)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# ۱. تعریف مدل SVM با هسته RBF (مناسب برای داده‌های هلالی شکل)
model = SVC(kernel='rbf', random_state=42)

# ۲. آموزش مدل با داده‌های Train
model.fit(X_train, y_train)

# ۳. پیش‌بینی روی داده‌های Test
y_pred = model.predict(X_test)

# ۴. محاسبه و نمایش نتایج
print("Accuracy (دقت مدل):", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix (ماتریس درهم‌ریختگی):\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report (گزارش کامل):\n", classification_report(y_test, y_pred))
```
