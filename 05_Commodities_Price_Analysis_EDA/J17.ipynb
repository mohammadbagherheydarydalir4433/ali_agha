```python
mas=np.random.rand(len(bit)) > 0.8
train = bit[mas]
test = bit[~mas]
```

```python
plt.scatter(train.Bronze_USD_Per_Ton, train.Gold_USD_Per_Oz, color='black')
plt.xlabel("Bronze_USD_Per_Ton")
plt.ylabel("Gold_USD_Per_Oz")
plt.show()
```

```python
from sklearn import linear_model
coin= linear_model.LinearRegression()
train_x=np.asanyarray(train[['Bronze_USD_Per_Ton']])
train_y=np.asanyarray(train[['Gold_USD_Per_Oz']])
coin.fit(train_x,train_y)


print('coefficients:' , coin.coef_)
print('intercept:' , coin.intercept_)
```

```python
plt.scatter(train.Bronze_USD_Per_Ton, train.Gold_USD_Per_Oz, color='red')
plt.plot(train_x, coin.coef_[0][0]* train_x, + coin.intercept_[0],'-r')
plt.xlabel("Bronze_USD_Per_Ton")
plt.ylabel("Gold_USD_Per_Oz")
```

```python
from sklearn.metrics import r2_score

test_x = np.asanyarray(test[['Bronze_USD_Per_Ton']])
test_y = np.asanyarray(test[['Gold_USD_Per_Oz']])
test_y_ = coin.predict(test_x)

print("Mean absolute error: %.2f" % np.mean(np.absolute(test_y_ - test_y)))
print("Residual sum of squares (MSE): %.2f" % np.mean((test_y_ - test_y) ** 2))
print("R2-score: %.2f" % r2_score(test_y , test_y_) )
```
